```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
import os

# آدرس یکی از عکس‌ها و ماسک‌های خروجی (مثلاً بیمار India_009)
img_path = "data/processed/aligned_raw/images"
mask_path = "data/processed/aligned_raw/masks"

# خواندن عکس و ماسک
image = cv2.imread(img_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

# ۱. بررسی ابعاد (Resolution)
print(f"📐 ابعاد عکس اصلی: {image.shape}")
print(f"📐 ابعاد ماسک تراز شده: {mask.shape}")
print("-" * 50)

# ایجاد یک تصویر ترکیبی (Overlay) برای بررسی چشمی
# ماسک را قرمز رنگ می‌کنیم تا به وضوح روی عکس مشخص شود
colored_mask = np.zeros_like(image)
colored_mask[:, :, 0] = mask  # اعمال ماسک روی کانال رنگ قرمز (Red Channel)

# ترکیب عکس و ماسک با شفافیت (آلفا بِلِندینگ)
overlay = cv2.addWeighted(image, 0.7, colored_mask, 0.6, 0)

# ۲. نمایش خروجی‌ها برای بررسی چشمی
fig, axes = plt.subplots(1, 3, figsize=(18, 6))

axes[0].imshow(image)
axes[0].set_title("1. Original Raw Image\n(Check for natural proportions)")
axes[0].axis("off")

axes[1].imshow(mask, cmap='gray')
axes[1].set_title("2. Aligned Mask\n(Warped via SIFT to fit)")
axes[1].axis("off")

axes[2].imshow(overlay)
axes[2].set_title("3. Overlay\n(Red area should perfectly cover the Palpebral Conjunctiva)")
axes[2].axis("off")

plt.tight_layout()
plt.show()
```
